```python
import json

# Load JSON file
with open("raw_links.json", "r", encoding="utf-8") as f:
    links = json.load(f)

# Filter links that end with the target string
target_suffix = "exam-aws-certified-developer-associate-dva-c02-topic-1/"
filtered_links = [link for link in links if link.endswith(target_suffix)]

# Deduplicate by converting to a set and back to list
unique_links = list(set(filtered_links))

# Optional: sort the links
unique_links.sort()

# Check the results
print(unique_links)

# Save filtered and deduplicated links back to JSON
with open("filtered_links.json", "w", encoding="utf-8") as f:
    json.dump(unique_links, f, ensure_ascii=False, indent=4)
```

```python
from pathlib import Path

folder = Path(r"output")  # <- change this

prefix = "new_"
count = 0

for p in folder.iterdir():
    if p.is_file() and p.suffix.lower() == ".pdf":
        if p.name.startswith(prefix):
            continue  # already prefixed
        new_name = p.with_name(prefix + p.name)
        if new_name.exists():
            print(f"SKIP (would overwrite): {new_name.name}")
            continue
        try:
            p.rename(new_name)
            print(f"RENAMED: {p.name} -> {new_name.name}")
            count += 1
        except Exception as e:
            print(f"ERROR renaming {p.name}: {e}")

print(f"Done. Renamed {count} file(s).")
```
